```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('input_videos/1.png', cv2.IMREAD_COLOR)
print(img.shape)
cropped = img[810:950 , 680:920]
cv2.imwrite('output_videos/cropped.jpg', cropped)
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
# Convert the image to HSV color space
# hsv = cv2.cvtColor(cropped, cv2.COLOR_BGR2HSV)

# Define the lower and upper bounds for yellow color
# lower_yellow = np.array([20, 90,90]) shafafa
# upper_yellow = np.array([30, 100, 100])

lower_yellow = np.array([20, 190, 240])
upper_yellow = np.array([30, 200, 255])



# Create a mask for yellow color
mask = cv2.inRange(cropped, lower_yellow, upper_yellow)


# Extract yellow regions using the mask
yellow_extracted = cv2.bitwise_and(cropped, cropped, mask=mask)
plt.imshow(cv2.cvtColor(yellow_extracted, cv2.COLOR_BGR2RGB))

```
